```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
!pip install matplotlib
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

print("✅ All imports loaded")
```

```python
t_set=tf.keras.utils.image_dataset_from_directory(
    "train",
    labels="inferred",
    label_mode="int",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(160, 160),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    format="tf",
    verbose=True,
)
```

```python
v_set=tf.keras.utils.image_dataset_from_directory(
    "val",
    labels="inferred",
    label_mode="int",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(160, 160),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    format="tf",
    verbose=True,
)
```

```python
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
IMG_SIZE = (160, 160)
class_names = t_set.class_names
num_classes = len(class_names)

num_classes = len(class_names)

data_aug = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2),
])

base_model = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3)
)

# 🔹 Initial training (frozen)
base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
x = data_aug(inputs)
x = tf.keras.applications.efficientnet.preprocess_input(x)

# ❗ FIX: removed training=False
x = base_model(x)

x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)  # 🔹 added
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(num_classes, activation="softmax", dtype="float32")(x)

model = models.Model(inputs, outputs)

# 🔹 Initial compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

callbacks = [
    EarlyStopping(monitor="val_accuracy", patience=3, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", patience=2, factor=0.2, min_lr=1e-6),
    ModelCheckpoint("best_model.keras", monitor="val_accuracy", mode="max", save_best_only=True)
]

# 🔹 Initial training
history = model.fit(t_set, validation_data=v_set, epochs=20, callbacks=callbacks)
```

```python
model.summary()
```

```python
# Unfreeze model
base_model.trainable = True

# Freeze all layers first
for layer in base_model.layers:
    layer.trainable = False

# Unfreeze last 40 layers
for layer in base_model.layers[-40:]:
    layer.trainable = True

# Freeze BatchNorm layers
for layer in base_model.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

# Recompile
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Fine-tune
history_finetune = model.fit(
    t_set,
    validation_data=v_set,
    epochs=8,
    callbacks=callbacks
)
```

```python
train_loss,train_accuracy=model.evaluate(v_set)
```

```python
model.save("trained_model.keras")
```

```python
import matplotlib.pyplot as plt

def plot_two_histories(h1, h2):
    acc = h1.history["accuracy"] + h2.history["accuracy"]
    val_acc = h1.history["val_accuracy"] + h2.history["val_accuracy"]
    loss = h1.history["loss"] + h2.history["loss"]
    val_loss = h1.history["val_loss"] + h2.history["val_loss"]

    epochs = range(1, len(acc)+1)

    plt.figure(figsize=(7,5))
    plt.plot(epochs, acc, label="Train Acc")
    plt.plot(epochs, val_acc, label="Val Acc")
    plt.title("Accuracy (Phase1 + FineTune)")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

    plt.figure(figsize=(7,5))
    plt.plot(epochs, loss, label="Train Loss")
    plt.plot(epochs, val_loss, label="Val Loss")
    plt.title("Loss (Phase1 + FineTune)")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.show()

plot_two_histories(history, history_finetune)
```
